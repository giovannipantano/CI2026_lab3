Child = list of C booleans
There are N children that have to be connected in a ring shape

Fitness = differences between charateristics of adiacent children
Compute the difference between one child and the next one7

A possible solution is a N*C matrix

A mutation can be swapping two random children
Cross over can be combining two children keeping only the distinct charateristics

```python
from collections.abc import Sequence
from random import sample
from itertools import permutations
from functools import total_ordering
```

```python
NUM_CHILDREN = 10
NUM_C = 5
```

```python
class Child:
    _name = str
    _carateristics = tuple[bool]

    def __init__(self, name: str, carateristics: Sequence[bool]):
        self._name = name
        self._carateristics = tuple(carateristics)

    def difference(self, other: Child) -> int:
        return sum(s != o for s, o in zip(self._carateristics, other._carateristics))

    def __str__(self) -> str:
        return f"{self._name}: {self._carateristics}"

    def __repr__(self) ->str:
        return f"{self._name}"

    def __lt__(self, other):
        return id(self) < id(other)

    def __eq__(self, other):
        return id(self) == id(other)
```

```python
children = list()
for c in range(NUM_CHILDREN):
    name = f"child {c}"
    children.append(Child(name, sample([True, False], k=NUM_C, counts=[NUM_C, NUM_C])))
```

```python
for c in children:
    print(c)
```

```python
def fitness(solution: Sequence[Child]):
    sol = list(solution)
    total_diff = 0
    n = len(solution)
    for i in range(n):
        c1 = sol[i]
        c2 = sol[(i+1) % n]
        total_diff += c1.difference(c2)

    return total_diff
```

```python
fitness(children)
```

```python
def tweak(sol):
    mutated = sol.copy()
    i, j = sample(range(1, len(sol)), 2)
    mutated[i], mutated[j] = mutated[j], mutated[i]
    return normalize(mutated)
```

```python
tweak(children)
```

```python
def normalize(sol):
    if sol[0] > sol[1]:
        return sol[::-1]
    else:
        return sol
```
